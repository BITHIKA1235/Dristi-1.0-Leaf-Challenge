```python
import pandas as pd
from sklearn.metrics import accuracy_score, f1_score
for n in ["1", "2"]:
    p = pd.read_csv(f"leaffinders_prediction_test{n}.csv")
    t = pd.read_csv(f"demo_test{n}_true_label.csv")
    m = p.merge(t, on="file")
    print(n, "matched", len(m), "acc", accuracy_score(m["label"], m["predicted"]),
          "F1", f1_score(m["label"], m["predicted"], average="macro"))
```
